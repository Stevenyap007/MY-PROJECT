```python
import os
import shutil
from glob import glob

# Original and destination paths
fitz_src = "Fitzpatrick17k_CategorizedAbbrvs\Categorized_AbbrvName"     # original dataset
fitz_dst = "fitz_selected"      # destination with ISIC-style class names

# Map ISIC class names to possible Fitz folders
class_map = {
    "mel": ["ma-me_111", "me_261", "su-sp-me-ss_118","le-ma_83"],
    "nv": ["co-ne_68","ha-ne_82","ne-ne_86"],
    "bcc": ["ba-ce-ca_468", "ba-ce-ca-mo_62", "so-cy-ba-ce-ca_66"],
    "scc": ["sq-ce-ca_581"],
    "ak": ["ac-ke_175"],
    "df": ["defi_79"],
    "bkl": ["se-ke_69"]
}

# Limit number of images per class
images_per_class = 500  # or match ISIC distribution

# Copy images to folders named exactly like ISIC class labels
for isic_class, folders in class_map.items():
    dst_dir = os.path.join(fitz_dst, isic_class)
    os.makedirs(dst_dir, exist_ok=True)

    copied = 0
    for folder in folders:
        src_folder = os.path.join(fitz_src, folder)
        if not os.path.exists(src_folder):
            continue
        for img_path in glob(os.path.join(src_folder, "*.*")):
            if copied >= images_per_class:
                break
            if img_path.lower().endswith((".jpg", ".jpeg", ".png")):
                shutil.copy(img_path, os.path.join(dst_dir, os.path.basename(img_path)))
                copied += 1
        if copied >= images_per_class:
            break
    print(f"Copied {copied} images to class folder '{isic_class}'")
```
