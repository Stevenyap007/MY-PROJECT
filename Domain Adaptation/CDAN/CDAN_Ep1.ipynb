```python
import torch
from cdan_mobilenetv2 import MobileNetV2CDAN

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

num_classes = 7
model = MobileNetV2CDAN(num_classes)
model.load_state_dict(torch.load("cdan_mobilenetv2.pth", map_location=device))
model = model.to(device)     
model.eval()
```

```python
from torch.utils.data import DataLoader
from torchvision import transforms
from torchvision.datasets import ImageFolder

transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize([0.485,0.456,0.406],
                         [0.229,0.224,0.225])
])

test_dataset = ImageFolder("/content/drive/MyDrive/Colab Notebooks/Domain Adaptation/fitz_selected", transform=transform)
test_loader = DataLoader(test_dataset, batch_size=32)

correct, total = 0, 0

with torch.no_grad():
    for x, y in test_loader:
        y_pred, _ = model(x)
        preds = y_pred.argmax(1)
        correct += (preds == y).sum().item()
        total += y.size(0)

print("Target accuracy:", correct / total)
```

```python
from torchvision.datasets import ImageFolder
from torchvision import transforms
from torch.utils.data import DataLoader

transform = transforms.Compose([
    transforms.Resize((224,224)),
    transforms.ToTensor(),
    transforms.Normalize(
        [0.485,0.456,0.406],
        [0.229,0.224,0.225]
    )
])

dataset = ImageFolder("/content/drive/MyDrive/Colab Notebooks/Domain Adaptation/fitz_selected", transform=transform)
loader = DataLoader(dataset, batch_size=32, shuffle=False)

class_names = dataset.classes
```

```python
import torch
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
```

```python
import numpy as np

y_true, y_pred = [], []

with torch.no_grad():
    for x, y in loader:
        x = x.to(device)
        logits, _ = model(x)
        preds = logits.argmax(1).cpu().numpy()

        y_true.extend(y.numpy())
        y_pred.extend(preds)

y_true = np.array(y_true)
y_pred = np.array(y_pred)
```

```python
from sklearn.metrics import classification_report

print(
    classification_report(
        y_true,
        y_pred,
        target_names=class_names,
        digits=4
    )
)
```

```python
from sklearn.metrics import classification_report
import pandas as pd

report_dict = classification_report(
    y_true,
    y_pred,
    target_names=class_names,
    output_dict=True
)

df = pd.DataFrame(report_dict).transpose().round(4)
df
```

```python
from torchvision.datasets import ImageFolder
from torchvision import transforms
from torch.utils.data import DataLoader

transform = transforms.Compose([
    transforms.Resize((224,224)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485,0.456,0.406],
        std=[0.229,0.224,0.225]
    )
])

dataset = ImageFolder("/content/drive/MyDrive/Colab Notebooks/Domain Adaptation/fitz_selected", transform=transform)
loader = DataLoader(dataset, batch_size=1, shuffle=True)

class_names = dataset.classes
```

```python
import matplotlib.pyplot as plt

samples = {}
with torch.no_grad():
    for img, label in loader:
        img = img.to(device)
        label = label.item()

        logits, _ = model(img)
        pred = logits.argmax(1).item()

        if pred == label and label not in samples:
            samples[label] = (img.cpu(), pred)

        if len(samples) == len(class_names):
            break

print("Collected samples:", samples.keys())
```

```python
import numpy as np

def denormalize(img):
    mean = torch.tensor([0.485, 0.456, 0.406]).view(3,1,1)
    std = torch.tensor([0.229, 0.224, 0.225]).view(3,1,1)
    return img * std + mean

plt.figure(figsize=(15,8))

for i, (cls, (img, pred)) in enumerate(samples.items()):
    plt.subplot(2, 4, i+1)

    img = denormalize(img.squeeze()).permute(1,2,0).numpy()
    plt.imshow(np.clip(img, 0, 1))
    plt.title(f"GT: {class_names[cls]}\nPred: {class_names[pred]}")
    plt.axis("off")

plt.suptitle("Sample Predictions per Class (Target Domain)")
plt.show()
```
